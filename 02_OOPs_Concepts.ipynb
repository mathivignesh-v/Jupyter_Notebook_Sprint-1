## 1. Class

### Step 1 – Understand the Concept

A **class** is a structure I create to describe what an object should contain and what it should be able to do. It groups related data and functions together. I can use the same class to create many objects with the same general structure.

### Step 2 – Demonstrate the Concept

A useful analogy is a **blueprint for a house**. The blueprint describes the structure of the house, but it is not the actual house. In the same way, a class describes the structure and behavior that objects created from it can have.

### Step 3 – Implement the Concept

The `Employee` class below contains employee-related behavior. `display()` is a method defined inside the class.

This demonstrates the concept because the class provides the structure from which an object can be created.

```python
class Employee:
    def display(self):
        return "Employee details"


employee = Employee()
print(employee.display())
```

## 2. Object

### Step 1 – Understand the Concept

An **object** is a usable instance created from a class. The class describes the structure, while the object is the actual entity I work with in the program. Multiple objects can be created from one class.

### Step 2 – Demonstrate the Concept

If a class is a blueprint for a house, an object is an actual house built from that blueprint. Two houses can follow the same blueprint while still being separate objects.

### Step 3 – Implement the Concept

The `Employee` class is used to create two separate objects. Each object is an instance of the same class.

This demonstrates that one class can be used to create multiple objects.

```python
class Employee:
    def __init__(self, name):
        self.name = name


employee1 = Employee("Vignesh")
employee2 = Employee("Arun")

print(employee1.name)
print(employee2.name)
```

## 3. Constructor (`__init__`)

### Step 1 – Understand the Concept

`__init__()` is a special method that runs automatically when I create an object. I use it to give the object its initial data. It helps make sure an object starts with the information it needs.

### Step 2 – Demonstrate the Concept

Think about registering a new employee in a company system. At the time of registration, the employee's name and role need to be entered. The constructor performs this initial setup when the object is created.

### Step 3 – Implement the Concept

The constructor receives `name` and `role` and stores them in the object.

This demonstrates automatic initialization because the method runs when `Employee()` is called.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role


employee = Employee("Vignesh", "AI/ML Engineer")

print(employee.name)
print(employee.role)
```

## 4. Instance Variables

### Step 1 – Understand the Concept

**Instance variables** store data that belongs to a particular object. Each object can have different values for the same instance variable because the values are stored separately for each object.

### Step 2 – Demonstrate the Concept

Imagine two employees. Both have a `name` and `role`, but their actual names and roles are different. These individual details belong to each employee object.

### Step 3 – Implement the Concept

`name` and `role` are assigned using `self`, so each object gets its own values.

Changing one employee's data does not automatically change the other employee's data.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role


employee1 = Employee("Vignesh", "Developer")
employee2 = Employee("Arun", "Tester")

employee1.role = "AI/ML Engineer"

print(employee1.name, "-", employee1.role)
print(employee2.name, "-", employee2.role)
```

## 5. Class Variables

### Step 1 – Understand the Concept

A **class variable** is shared by all objects created from a class. I use it when the value is common to every instance instead of being different for each object.

### Step 2 – Demonstrate the Concept

Suppose every employee belongs to the same company. The company name can be stored once at the class level instead of storing a separate copy for every employee.

### Step 3 – Implement the Concept

`company` is defined inside the class but outside any method. Both objects can access the same class variable.

```python
class Employee:
    company = "Tech Solutions"

    def __init__(self, name):
        self.name = name


employee1 = Employee("Vignesh")
employee2 = Employee("Arun")

print(employee1.company)
print(employee2.company)
```

## 6. Instance Methods

### Step 1 – Understand the Concept

An **instance method** works with a particular object. Its first parameter is normally `self`, which gives the method access to that object's instance variables.

### Step 2 – Demonstrate the Concept

An employee object can perform an action such as displaying its own details. The method needs access to that employee's data, so it works with the specific instance.

### Step 3 – Implement the Concept

`display_details()` is an instance method because it uses `self.name` and `self.role`.

Each object can call the method and receive its own details.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def display_details(self):
        return f"{self.name} works as {self.role}"


employee = Employee("Vignesh", "AI/ML Engineer")
print(employee.display_details())
```

## 7. Static Methods

### Step 1 – Understand the Concept

A **static method** belongs logically to a class but does not need access to a particular object or the class itself. I use `@staticmethod` when the operation is related to the class but can work independently using its arguments.

### Step 2 – Demonstrate the Concept

A company may have a utility calculation such as checking whether an age is valid. The calculation does not need employee-specific data, so it can be a static method.

### Step 3 – Implement the Concept

`is_adult()` does not use `self` or `cls`. It only uses the value passed to it.

That is what makes it suitable as a static method.

```python
class Employee:
    @staticmethod
    def is_adult(age):
        return age >= 18


print(Employee.is_adult(22))
print(Employee.is_adult(16))
```

## 8. Class Methods

### Step 1 – Understand the Concept

A **class method** works with the class itself rather than a particular object. It uses `cls` as its first parameter and can access or modify class-level data.

### Step 2 – Demonstrate the Concept

If the company name is common to all employees, a class method can be used to update that shared company information.

### Step 3 – Implement the Concept

`change_company()` changes the class variable `company`. The change is visible through objects created from the class as well.

```python
class Employee:
    company = "Tech Solutions"

    @classmethod
    def change_company(cls, new_name):
        cls.company = new_name


print(Employee.company)

Employee.change_company("Innovative Systems")

print(Employee.company)
```

## 9. Properties

### Step 1 – Understand the Concept

A **property** lets me access a method like an attribute. It is useful when I want to control how a value is read or changed while keeping simple attribute-style access.

### Step 2 – Demonstrate the Concept

Think of an employee's salary. I may want to allow the salary to be read normally but prevent invalid values from being assigned.

### Step 3 – Implement the Concept

The `salary` property uses a getter to return the value and a setter to validate a new value.

The validation happens automatically when `employee.salary` is assigned.

```python
class Employee:
    def __init__(self, salary):
        self._salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value


employee = Employee(30000)
employee.salary = 35000

print(employee.salary)
```

## 10. Encapsulation

### Step 1 – Understand the Concept

**Encapsulation** means keeping related data and behavior together and controlling how important data is accessed or changed. In Python, naming conventions and properties are commonly used to indicate and control internal data.

### Step 2 – Demonstrate the Concept

A bank account should not allow any part of the program to change the balance carelessly. Instead, operations such as deposit and withdrawal can control how the balance changes.

### Step 3 – Implement the Concept

The account balance is stored as `_balance`, and changes happen through `deposit()` and `withdraw()`.

This demonstrates encapsulation because the class controls the operations performed on its internal state.

```python
class BankAccount:
    def __init__(self, balance):
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount

    def get_balance(self):
        return self._balance


account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)

print(account.get_balance())
```

## 11. Abstraction

### Step 1 – Understand the Concept

**Abstraction** means focusing on the important interface while hiding unnecessary implementation details. The user of a class can use the required operation without needing to know every internal step.

### Step 2 – Demonstrate the Concept

When I use an ATM, I choose an action such as withdrawing money. I do not need to know the internal banking and network operations that happen behind the screen.

### Step 3 – Implement the Concept

The abstract `Payment` class defines the `pay()` operation. The concrete class provides the actual implementation.

This demonstrates that the user interacts with the required operation without needing the internal implementation details.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class UPI(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using UPI"


payment = UPI()
print(payment.pay(500))
```

## 12. Inheritance

### Step 1 – Understand the Concept

**Inheritance** allows one class to reuse attributes and methods from another class. The new class can also add its own behavior or modify inherited behavior.

### Step 2 – Demonstrate the Concept

A general `Employee` class may contain common information such as a name. A `Developer` class can inherit that information and add developer-specific behavior.

### Step 3 – Implement the Concept

`Developer` inherits from `Employee`. It automatically gets the `name` attribute and `display_name()` method while adding `write_code()`.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        return self.name


class Developer(Employee):
    def write_code(self):
        return "Writing Python code"


developer = Developer("Vignesh")

print(developer.display_name())
print(developer.write_code())
```

## 13. Types of Inheritance

### Step 1 – Understand the Concept

Python supports several inheritance structures. Common types are **single**, **multiple**, **multilevel**, and **hierarchical** inheritance. The structure describes how classes are related to one another.

### Step 2 – Demonstrate the Concept

A simple example is a multilevel structure: an `Employee` can be a `Developer`, and a specialized `PythonDeveloper` can inherit from `Developer`. Each level adds or reuses behavior.

### Step 3 – Implement the Concept

The example below demonstrates **multilevel inheritance**. `PythonDeveloper` inherits from `Developer`, which inherits from `Employee`.

The final object can use methods from all levels of the hierarchy.

```python
class Employee:
    def work(self):
        return "Employee is working"


class Developer(Employee):
    def code(self):
        return "Developer is coding"


class PythonDeveloper(Developer):
    def python_code(self):
        return "Writing Python code"


developer = PythonDeveloper()

print(developer.work())
print(developer.code())
print(developer.python_code())
```

## 14. Polymorphism

### Step 1 – Understand the Concept

**Polymorphism** means the same operation or method name can work differently depending on the object using it. This allows code to work with different object types through a common interface.

### Step 2 – Demonstrate the Concept

A `make_sound()` operation can produce different sounds for a dog and a cat. The operation has the same purpose, but each object provides its own behavior.

### Step 3 – Implement the Concept

Both classes define `make_sound()`. The loop calls the same method on different objects.

This demonstrates polymorphism because the result depends on the object being used.

```python
class Dog:
    def make_sound(self):
        return "Woof"


class Cat:
    def make_sound(self):
        return "Meow"


animals = [Dog(), Cat()]

for animal in animals:
    print(animal.make_sound())
```

## 15. Method Overriding

### Step 1 – Understand the Concept

**Method overriding** happens when a child class provides its own version of a method that already exists in the parent class. The child version is used when the method is called on the child object.

### Step 2 – Demonstrate the Concept

A general employee may have a `work()` method. A developer can override it to describe developer-specific work.

### Step 3 – Implement the Concept

Both classes define `work()`, but `Developer` provides a different implementation.

Calling `work()` on the developer object uses the overridden method.

```python
class Employee:
    def work(self):
        return "Employee is working"


class Developer(Employee):
    def work(self):
        return "Developer is writing code"


employee = Employee()
developer = Developer()

print(employee.work())
print(developer.work())
```

## 16. Method Overloading in Python

### Step 1 – Understand the Concept

Python does not support traditional method overloading by defining multiple methods with the same name and different parameter lists. A common Python approach is to use default arguments or variable-length arguments to handle different numbers of inputs.

### Step 2 – Demonstrate the Concept

A calculator may need to add two numbers or three numbers. Instead of defining multiple `add()` methods, one method can accept a variable number of values.

### Step 3 – Implement the Concept

The `*numbers` parameter allows the same `add()` method to accept different numbers of arguments.

This provides overloading-like behavior using Python's flexible argument handling.

```python
class Calculator:
    def add(self, *numbers):
        return sum(numbers)


calculator = Calculator()

print(calculator.add(10, 20))
print(calculator.add(10, 20, 30, 40))
```

## 17. Composition

### Step 1 – Understand the Concept

**Composition** represents a strong "has-a" relationship where one object contains another object as an important part of itself. The contained object is created as part of the larger object's structure.

### Step 2 – Demonstrate the Concept

A car has an engine. The car uses the engine as one of its components, so the engine is part of the car's design.

### Step 3 – Implement the Concept

`Car` creates and stores an `Engine` object. The car then uses that engine through its own `start()` method.

This demonstrates a has-a relationship through object composition.

```python
class Engine:
    def start(self):
        return "Engine started"


class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        return self.engine.start()


car = Car()
print(car.start())
```

## 18. Aggregation

### Step 1 – Understand the Concept

**Aggregation** is a "has-a" relationship where one object uses another object, but the contained object can exist independently. The lifetime of the two objects is not necessarily tied together.

### Step 2 – Demonstrate the Concept

A department has employees, but an employee can continue to exist even if the department object is removed. The employee is not completely dependent on that department object.

### Step 3 – Implement the Concept

The employees are created outside the `Department` class and then passed into it.

This demonstrates that the department uses existing objects rather than creating their entire lifetime internally.

```python
class Employee:
    def __init__(self, name):
        self.name = name


class Department:
    def __init__(self, employees):
        self.employees = employees

    def show_employees(self):
        return [employee.name for employee in self.employees]


employee1 = Employee("Vignesh")
employee2 = Employee("Arun")

department = Department([employee1, employee2])

print(department.show_employees())
```

## 19. Association

### Step 1 – Understand the Concept

**Association** represents a relationship between independent objects. The objects can interact with each other without one necessarily owning the other.

### Step 2 – Demonstrate the Concept

A teacher and a student can interact in a classroom. Both can exist independently, but they have a relationship through teaching and learning.

### Step 3 – Implement the Concept

The `Teacher` method receives a `Student` object as an argument.

Neither class creates or owns the other, which demonstrates a simple association.

```python
class Student:
    def __init__(self, name):
        self.name = name


class Teacher:
    def teach(self, student):
        return f"Teaching {student.name}"


student = Student("Vignesh")
teacher = Teacher()

print(teacher.teach(student))
```

## 20. Magic (Dunder) Methods

### Step 1 – Understand the Concept

**Magic methods**, also called **dunder methods**, have names surrounded by double underscores, such as `__init__` and `__str__`. Python calls these methods automatically in specific situations.

### Step 2 – Demonstrate the Concept

When I use `print(object)`, Python needs to know how the object should be represented as text. Implementing `__str__()` lets me define that behavior.

### Step 3 – Implement the Concept

The `__str__()` method returns a readable description of the object.

When `print(employee)` is used, Python automatically calls `__str__()`.

```python
class Employee:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def __str__(self):
        return f"{self.name} - {self.role}"


employee = Employee("Vignesh", "AI/ML Engineer")
print(employee)
```

## 21. `super()`

### Step 1 – Understand the Concept

`super()` gives a child class a convenient way to access methods or the constructor of its parent class. It is especially useful when the child wants to reuse parent initialization and then add its own setup.

### Step 2 – Demonstrate the Concept

A developer is still an employee, so the developer can reuse the employee's initialization for the name and then add a programming language.

### Step 3 – Implement the Concept

`Developer.__init__()` uses `super().__init__(name)` to reuse the parent constructor.

After that, it adds the `language` attribute specific to the developer.

```python
class Employee:
    def __init__(self, name):
        self.name = name


class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language


developer = Developer("Vignesh", "Python")

print(developer.name)
print(developer.language)
```

## 22. Dataclasses

### Step 1 – Understand the Concept

A **dataclass** is a convenient way to create classes that mainly store data. Python can automatically generate common methods such as `__init__()` and a useful string representation.

### Step 2 – Demonstrate the Concept

If I only need a class to represent a student record, writing all the basic initialization code manually can be repetitive. A dataclass reduces that boilerplate.

### Step 3 – Implement the Concept

`@dataclass` automatically provides the constructor for `Student`.

The class can then be created directly using the declared fields.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    course: str


student = Student("Vignesh", 22, "Python")

print(student)
```

## 23. Singleton Pattern

### Step 1 – Understand the Concept

The **Singleton pattern** is a design approach where a class is intended to have only one shared instance. It can be useful when an application needs one shared object, such as a configuration manager.

### Step 2 – Demonstrate the Concept

Imagine one central configuration object used by different parts of an application. Instead of creating separate configuration objects, the application can reuse one shared instance.

### Step 3 – Implement the Concept

The `__new__()` method controls object creation. The class stores the first instance in `_instance` and returns that same instance for later requests.

This demonstrates the main idea of a singleton: repeated construction returns the same object.

```python
class AppConfig:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


config1 = AppConfig()
config2 = AppConfig()

print(config1 is config2)
```
