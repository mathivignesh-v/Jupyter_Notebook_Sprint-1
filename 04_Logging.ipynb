## 1. What is Logging?

### Step 1 – Understand the Concept

Logging is a way for a program to leave a useful record of what it is doing. Instead of relying only on temporary screen output, I can record events such as startup, warnings, failures, and debugging details.

### Step 2 – Demonstrate the Concept

A useful analogy is a diary for an application. The program writes down important events as they happen, so I can look back later and understand what happened.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("An error occurred")
```

### Implementation Explanation

This creates basic log messages with the built-in `logging` module. The calls to `info()`, `warning()`, and `error()` demonstrate recording different events.

## 2. Why Use Logging?

### Step 1 – Understand the Concept

Logging becomes useful when a program is larger or runs without someone watching the console. It gives me a structured history that I can use for troubleshooting and monitoring.

### Step 2 – Demonstrate the Concept

For example, if a website fails at night, I cannot ask the application what happened. A log file can provide the sequence of events that led to the failure.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

logging.info("User logged in")
logging.warning("Password attempt limit is almost reached")
```

### Implementation Explanation

The logger records normal activity and a warning. This demonstrates how logging can preserve useful application information instead of depending only on `print()`.

## 3. `print()` vs `logging`

### Step 1 – Understand the Concept

`print()` is useful for quick output while learning or checking a small piece of code. Logging is designed for application diagnostics because it supports severity levels, formatting, handlers, and configuration.

## 4. Log Levels

### Step 1 – Understand the Concept

Log levels tell me how important a message is. I can use DEBUG for detailed troubleshooting, INFO for normal events, WARNING for situations needing attention, ERROR for failures, and CRITICAL for serious failures.

### Step 2 – Demonstrate the Concept

Think of levels as priority labels on tasks: a normal update is low priority, a warning needs attention, an error means something failed, and a critical message signals an urgent problem.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(levelname)s: %(message)s"
)

logging.debug("Debug information")
logging.info("Application is running")
logging.warning("This is a warning")
logging.error("An error occurred")
logging.critical("Critical problem detected")
```

### Implementation Explanation

The five logging calls demonstrate the available severity levels. Setting the logger to `DEBUG` allows all of these messages to be displayed.

## 10. Log Formatters

### Step 1 – Understand the Concept

A formatter controls the information and layout included in each log message. I can include time, logger name, level, and the actual message.

### Step 2 – Demonstrate the Concept

It is like choosing the columns in an incident report. Instead of writing only the event, I can include when it happened and how serious it was.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Formatted log message")
```

### Implementation Explanation

The formatter adds the time, level, and message to each entry. This makes the log easier to read and investigate.

## 11. File Handlers

### Step 1 – Understand the Concept

A file handler sends log records to a file. This is useful when I need logs to remain available after the program finishes.

### Step 2 – Demonstrate the Concept

It is like keeping a written register at the entrance of a building. Events can be reviewed later instead of disappearing when the screen is closed.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logger = logging.getLogger("my_application")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

logger.debug("Debug message")
logger.info("Application started")
logger.warning("Warning message")
```

### Implementation Explanation

`FileHandler` connects the logger to `application.log`. The formatter gives the saved entries a consistent structure.

## 12. Console Handlers

### Step 1 – Understand the Concept

A console handler sends log records to the console. This is useful when I want to watch what an application is doing while it is running.

### Step 2 – Demonstrate the Concept

It is like a live status board: I can see events immediately while the application is operating.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logger = logging.getLogger("file_logger")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

logger.info("This message is written to application.log")
logger.error("An example error was logged")
```

### Implementation Explanation

`StreamHandler` sends INFO and WARNING messages to the console and the formatter controls their appearance.

## 14. Rotating Logs – Bonus

### Step 1 – Understand the Concept

Rotating logs prevent a single log file from growing forever. A rotating handler creates backup files when the active log reaches a configured size.

### Step 2 – Demonstrate the Concept

This is similar to replacing a full notebook with a new notebook while keeping the older notebooks as backups. The history is retained without one file becoming huge.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logger = logging.getLogger("console_logger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

logger.info("This message appears in the console")
logger.warning("Warning displayed in the console")
```

### Implementation Explanation

`RotatingFileHandler` uses `maxBytes` to control file size and `backupCount` to keep older files. The loop generates enough messages to demonstrate rotation.

## 15. Logging Exceptions

### Step 1 – Understand the Concept

When an exception happens, logging should capture enough information to understand the failure. `logger.exception()` is useful inside an `except` block because it includes traceback details.

### Step 2 – Demonstrate the Concept

It is like recording not only that a machine stopped, but also the diagnostic information showing where the failure occurred.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging

logger = logging.getLogger("application")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()
file_handler = logging.FileHandler("application_full.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.info("Application started")
logger.warning("Example warning")
logger.error("Example error")
```

### Implementation Explanation

The division by zero is caught and logged with `logging.exception()`. The traceback helps identify the source of the problem.

## 16. Practical Example – User Login System

### Step 1 – Understand the Concept

A practical logging system combines levels according to the meaning of each event. A login attempt can produce DEBUG information, INFO for success, or WARNING for invalid credentials.

### Step 2 – Demonstrate the Concept

A security guard records an entry when someone arrives, a normal event when access is granted, and a warning when an access attempt fails.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_logger")
logger.setLevel(logging.INFO)

rotating_handler = RotatingFileHandler(
    "rotating_application.log",
    maxBytes=1024,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

rotating_handler.setFormatter(formatter)
logger.addHandler(rotating_handler)

for i in range(100):
    logger.info("Log message number %s", i)

print("Rotating log example completed.")
```

### Implementation Explanation

The program checks login details and records the result using appropriate log levels. This demonstrates logging in a realistic application flow.

### Step 1 – Understand the Concept

Good logging should be useful rather than noisy. I should use meaningful levels, clear messages, consistent formatting, and avoid sensitive information such as passwords or API keys.

### Step 2 – Demonstrate the Concept

The example/analogy above shows the main idea before moving to the implementation.
