## 1. `try`

### Step 1 – Understand the Concept

`try` marks code where an exception might happen. It gives Python a controlled place to watch for a runtime problem.

### Step 2 – Demonstrate the Concept

It is like trying an operation inside a safety area. If something goes wrong, the program can move to an appropriate handling step instead of failing unexpectedly.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
try:
    number = 10 / 2
    print(number)
except Exception:
    print("An error occurred")
```

### Implementation Explanation

The division is placed inside `try`. If it fails, the `except` block can handle it.

## 2. `except`

### Step 1 – Understand the Concept

`except` defines what the program should do when a particular exception occurs. Catching a specific exception makes the response clearer.

### Step 2 – Demonstrate the Concept

It is like having a separate response procedure for a known problem. A missing file and invalid number are different problems, so they can have different handlers.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("Invalid input. Please enter a number.")
```

### Implementation Explanation

The code catches `ValueError` and gives the user a useful message rather than exposing the raw error.

## 3. `else`

### Step 1 – Understand the Concept

`else` runs only when the `try` block completes without an exception. It is useful for code that should happen only after the risky operation succeeds.

### Step 2 – Demonstrate the Concept

It is like saying, 'If the check passes, continue with the next step.'

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
try:
    number = int("abc")
except ValueError:
    print("Invalid number")
```

### Implementation Explanation

The conversion happens in `try`, while the success message is placed in `else`.

## 4. `finally`

### Step 1 – Understand the Concept

`finally` runs whether an exception occurs or not. It is useful for cleanup work that should happen in every case.

### Step 2 – Demonstrate the Concept

It is like closing a shop at the end of the day whether or not there was a problem during business hours.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("Result:", a / b)
except ValueError:
    print("Please enter valid numbers")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Implementation Explanation

The final message runs after the `try`/`except` flow regardless of the result.

## 5. Combining `try`, `except`, `else`, and `finally`

### Step 1 – Understand the Concept

These blocks can be combined to separate risky work, error handling, successful processing, and cleanup.

### Step 2 – Demonstrate the Concept

## 19. Practical Example – Student Marks Validation

### Step 1 – Understand the Concept

This program combines exception handling techniques around a realistic input-validation task. The aim is to reject invalid marks while giving the user a useful result.

### Step 2 – Demonstrate the Concept

It is like a teacher checking a mark sheet: the value must be numeric, within an allowed range, and then accepted only if it passes the checks.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
class InvalidMarksError(Exception):
    pass

def validate_marks(marks):
    if marks < 0 or marks > 100:
        raise InvalidMarksError("Marks must be between 0 and 100")
    assert 0 <= marks <= 100, "Marks must be valid"
    return marks

try:
    student_name = input("Enter student name: ")
    marks = float(input("Enter marks: "))
    valid_marks = validate_marks(marks)
except ValueError:
    print("Please enter a valid number for marks.")
except InvalidMarksError as error:
    print("Error:", error)
except AssertionError as error:
    print("Assertion Error:", error)
else:
    print(f"{student_name} scored {valid_marks} marks.")
finally:
    print("Marks validation completed.")
```

### Implementation Explanation

The program uses a custom exception, `raise`, `assert`, multiple `except` blocks, `else`, and `finally` to validate student marks.

## 20. Practical Example – Calculator

### Step 1 – Understand the Concept

A calculator is a useful example because different invalid inputs can produce different exceptions. Exception handling keeps those cases separate.

### Step 2 – Demonstrate the Concept

It is like a calculator at a service desk: it should reject invalid numbers, reject division by zero, and only display a result when the operation is valid.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
def calculator(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            raise ZeroDivisionError("Cannot divide by zero")
        return a / b
    else:
        raise ValueError("Invalid operator")

try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")
    result = calculator(a, b, operator)
except ValueError as error:
    print("Value Error:", error)
except ZeroDivisionError as error:
    print("Math Error:", error)
else:
    print("Result:", result)
finally:
    print("Calculator execution completed.")
```

### Implementation Explanation

The calculator raises specific errors for invalid operators and division by zero, then handles them separately.
