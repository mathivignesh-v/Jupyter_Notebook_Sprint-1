# Part 1 – Database Design

## 1. Database

### Step 1 – Understand the Concept

A database is an organized place where related information is stored so that it can be searched, updated, and managed efficiently.

For example, a college application might store student details, courses, and marks in related tables instead of keeping everything in separate text files.

### Step 2 – Demonstrate the Concept

Think of a database like a well-organized cupboard. Instead of putting every item into one large box, related items are kept in separate sections.

For example:
- Students → student information
- Courses → course information
- Enrollments → which student joined which course

This organization makes it easier to find and manage information.

### Step 3 – Implement the Concept

The following statement creates a database in PostgreSQL:

```sql
CREATE DATABASE college_db;
```

The `CREATE DATABASE` statement demonstrates how a database can be created to hold related tables and data.

````python
CREATE DATABASE college_db;
````

## 2. Tables

### Step 1 – Understand the Concept

A table stores data in rows and columns. Each column represents a type of information, while each row represents one record.

### Step 2 – Demonstrate the Concept

A `students` table can be compared to a spreadsheet:

| student_id | name | age |
|---|---|---|
| 1 | Vignesh | 22 |
| 2 | Arun | 21 |

Each row represents a student and each column describes one property.

### Step 3 – Implement the Concept

```sql
CREATE TABLE students (
    student_id INTEGER,
    name VARCHAR(100),
    age INTEGER
);
```

`CREATE TABLE` defines the columns that will exist in the table.

````python
CREATE TABLE students (
    student_id INTEGER,
    name VARCHAR(100),
    age INTEGER
);
````

## 3. Primary Key

### Step 1 – Understand the Concept

A primary key is a column, or combination of columns, used to uniquely identify each row in a table. Two rows should not have the same primary-key value.

### Step 2 – Demonstrate the Concept

A student ID is a good example. If two students have the same name, their student IDs can still uniquely identify them.

### Step 3 – Implement the Concept

```sql
CREATE TABLE students_pk (
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(100),
    age INTEGER
);
```

`PRIMARY KEY` makes `student_id` the unique identifier for each student.

````python
CREATE TABLE students_pk (
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(100),
    age INTEGER
);
````

## 4. Foreign Key

### Step 1 – Understand the Concept

A foreign key connects a column in one table to a key in another table. It helps maintain a valid relationship between related records.

### Step 2 – Demonstrate the Concept

If a student enrolls in a course, the enrollment record can store the student's ID. That ID should refer to an existing student.

### Step 3 – Implement the Concept

```sql
CREATE TABLE enrollments (
    enrollment_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    FOREIGN KEY (student_id) REFERENCES students_pk(student_id)
);
```

The `FOREIGN KEY` connects `enrollments.student_id` with `students_pk.student_id`.

````python
CREATE TABLE enrollments (
    enrollment_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    FOREIGN KEY (student_id) REFERENCES students_pk(student_id)
);
````

## 5. Relationships

### Step 1 – Understand the Concept

A relationship describes how records in different tables are connected.

Common relationships include:
- One-to-one
- One-to-many
- Many-to-many

### Step 2 – Demonstrate the Concept

One student can enroll in many courses, while a course can contain many students. This is a many-to-many relationship and is usually represented using a junction table such as `enrollments`.

### Step 3 – Implement the Concept

```sql
CREATE TABLE courses (
    course_id INTEGER PRIMARY KEY,
    course_name VARCHAR(100)
);

CREATE TABLE student_courses (
    student_id INTEGER,
    course_id INTEGER,
    PRIMARY KEY (student_id, course_id),
    FOREIGN KEY (student_id) REFERENCES students_pk(student_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
);
```

The `student_courses` table connects students and courses and demonstrates a many-to-many relationship.

````python
CREATE TABLE courses (
    course_id INTEGER PRIMARY KEY,
    course_name VARCHAR(100)
);

CREATE TABLE student_courses (
    student_id INTEGER,
    course_id INTEGER,
    PRIMARY KEY (student_id, course_id),
    FOREIGN KEY (student_id) REFERENCES students_pk(student_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
);
````

## 6. Constraints

### Step 1 – Understand the Concept

Constraints are rules applied to table data. They help prevent invalid information from being stored.

Common constraints include:
- `PRIMARY KEY`
- `FOREIGN KEY`
- `NOT NULL`
- `UNIQUE`
- `CHECK`
- `DEFAULT`

### Step 2 – Demonstrate the Concept

Imagine a form where a student's name is mandatory and marks must be between 0 and 100. Constraints can make the database enforce those rules instead of relying only on application code.

### Step 3 – Implement the Concept

````python
CREATE TABLE student_results (
    result_id INTEGER PRIMARY KEY,
    student_name VARCHAR(100) NOT NULL,
    email VARCHAR(150) UNIQUE,
    marks INTEGER CHECK (marks BETWEEN 0 AND 100),
    status VARCHAR(20) DEFAULT 'Active'
);
````

The table demonstrates several constraints:
- `PRIMARY KEY` identifies each result.
- `NOT NULL` requires a student name.
- `UNIQUE` prevents duplicate email addresses.
- `CHECK` restricts marks to 0–100.
- `DEFAULT` supplies a value when status is omitted.

## 7. SELECT

### Step 1 – Understand the Concept

`SELECT` is used to retrieve data from one or more columns.

### Step 2 – Demonstrate the Concept

If a spreadsheet contains many columns but I only want to see employee names and departments, I can select only those columns.

### Step 3 – Implement the Concept

````python
SELECT name, department
FROM employees;
````

The query retrieves only the `name` and `department` columns from `employees`.

## 8. WHERE

### Step 1 – Understand the Concept

`WHERE` filters rows based on a condition.

### Step 2 – Demonstrate the Concept

If I have 100 employees but want only employees from the IT department, I need a condition to filter the records.

### Step 3 – Implement the Concept

````python
SELECT name, salary
FROM employees
WHERE department = 'IT';
````

The `WHERE` condition keeps only rows where the department is `IT`.

## 9. ORDER BY

### Step 1 – Understand the Concept

`ORDER BY` sorts query results.

### Step 2 – Demonstrate the Concept

A salary list can be arranged from lowest to highest or highest to lowest.

### Step 3 – Implement the Concept

````python
SELECT name, salary
FROM employees
ORDER BY salary DESC;
````

`DESC` sorts the salary from highest to lowest. `ASC` can be used for ascending order.

## 10. GROUP BY

### Step 1 – Understand the Concept

`GROUP BY` combines rows that have the same value in selected columns. It is commonly used with aggregate functions such as `COUNT`, `SUM`, and `AVG`.

### Step 2 – Demonstrate the Concept

Instead of looking at every employee separately, I might want to know how many employees belong to each department.

### Step 3 – Implement the Concept

````python
SELECT department, COUNT(*) AS employee_count
FROM employees
GROUP BY department;
````

The query creates one group for each department and counts the employees in each group.

## 11. HAVING

### Step 1 – Understand the Concept

`HAVING` filters grouped results. Unlike `WHERE`, which normally filters individual rows before grouping, `HAVING` filters groups after aggregation.

### Step 2 – Demonstrate the Concept

If I want only departments that have at least five employees, I need to filter the result of `COUNT(*)`.

### Step 3 – Implement the Concept

````python
SELECT department, COUNT(*) AS employee_count
FROM employees
GROUP BY department
HAVING COUNT(*) >= 5;
````

`HAVING` keeps only departments whose employee count is at least five.

## 12. DISTINCT

### Step 1 – Understand the Concept

`DISTINCT` removes duplicate values from the selected result.

### Step 2 – Demonstrate the Concept

If many employees live in the same city, selecting cities normally may repeat city names. `DISTINCT` gives each city once.

### Step 3 – Implement the Concept

````python
SELECT DISTINCT city
FROM employees;
````

The query returns each city only once.

## 13. LIMIT

### Step 1 – Understand the Concept

`LIMIT` restricts how many rows are returned.

### Step 2 – Demonstrate the Concept

If a table has thousands of records but I only want to see the first five results while exploring the data, I can use `LIMIT`.

### Step 3 – Implement the Concept

````python
SELECT name, salary
FROM employees
ORDER BY salary DESC
LIMIT 5;
````

The query returns only the five highest-paid employees after sorting.

# Part 3 – Joins

Assume these tables:

```text
employees(employee_id, name, department_id)
departments(department_id, department_name)
```

## 14. INNER JOIN

### Step 1 – Understand the Concept

`INNER JOIN` returns rows where matching records exist in both tables.

### Step 2 – Demonstrate the Concept

Imagine two lists: employees and departments. An inner join shows employees whose department ID can be matched to a department.

### Step 3 – Implement the Concept

````python
SELECT e.name, d.department_name
FROM employees AS e
INNER JOIN departments AS d
    ON e.department_id = d.department_id;
````

The `ON` condition tells SQL how the two tables are related. Only matching records are returned.

## 15. LEFT JOIN

### Step 1 – Understand the Concept

`LEFT JOIN` returns every row from the left table and matching rows from the right table. If there is no match, right-table columns contain `NULL`.

### Step 2 – Demonstrate the Concept

This is useful when I want to see every employee, including employees who have not yet been assigned to a department.

### Step 3 – Implement the Concept

````python
SELECT e.name, d.department_name
FROM employees AS e
LEFT JOIN departments AS d
    ON e.department_id = d.department_id;
````

All employees are preserved because `employees` is on the left side.

## 16. RIGHT JOIN

### Step 1 – Understand the Concept

`RIGHT JOIN` keeps every row from the right table and includes matching rows from the left table.

### Step 2 – Demonstrate the Concept

It can show every department, including departments that currently have no matching employee.

### Step 3 – Implement the Concept

````python
SELECT e.name, d.department_name
FROM employees AS e
RIGHT JOIN departments AS d
    ON e.department_id = d.department_id;
````

Every department is retained because `departments` is the right-side table.

## 17. FULL JOIN

### Step 1 – Understand the Concept

`FULL JOIN` returns matching rows plus unmatched rows from both tables.

### Step 2 – Demonstrate the Concept

It is useful when I want a complete comparison showing employees without departments and departments without employees.

### Step 3 – Implement the Concept

````python
SELECT e.name, d.department_name
FROM employees AS e
FULL OUTER JOIN departments AS d
    ON e.department_id = d.department_id;
````

Both sides are preserved. Where no matching row exists, the missing side is represented by `NULL`.

> MySQL does not directly support `FULL OUTER JOIN`; in MySQL this can be simulated with combinations of `LEFT JOIN`, `RIGHT JOIN`, and `UNION`.

## 18. CROSS JOIN

### Step 1 – Understand the Concept

`CROSS JOIN` creates every possible combination of rows from two tables.

### Step 2 – Demonstrate the Concept

If there are three employees and four training courses, a cross join produces 12 employee-course combinations.

### Step 3 – Implement the Concept

````python
SELECT e.name, c.course_name
FROM employees AS e
CROSS JOIN courses AS c;
````

No matching condition is required. Every employee is paired with every course.

# Part 4 – Advanced SQL

## 19. UNION

### Step 1 – Understand the Concept

`UNION` combines the results of two compatible `SELECT` queries and removes duplicate rows.

### Step 2 – Demonstrate the Concept

If I have employees from two branches stored in separate tables, I can combine their names into one result.

### Step 3 – Implement the Concept

````python
SELECT name
FROM employees_chennai

UNION

SELECT name
FROM employees_bangalore;
````

Both queries must return compatible numbers and types of columns. `UNION ALL` can be used when duplicates should also be retained.

## 20. CASE

### Step 1 – Understand the Concept

`CASE` allows conditional logic inside a SQL query. It is similar to an `if / elif / else` decision.

### Step 2 – Demonstrate the Concept

An employee earning 80,000 or more could be labeled `High`, while lower salaries could be labeled `Standard`.

### Step 3 – Implement the Concept

````python
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 80000 THEN 'High'
        WHEN salary >= 50000 THEN 'Medium'
        ELSE 'Standard'
    END AS salary_category
FROM employees;
````

The `CASE` expression checks salary conditions and creates a new calculated category.

## 21. Subqueries

### Step 1 – Understand the Concept

A subquery is a query written inside another query. The inner query provides a value or set of rows that the outer query uses.

### Step 2 – Demonstrate the Concept

To find employees earning more than the average salary, I first need the average salary and then compare each employee with it.

### Step 3 – Implement the Concept

````python
SELECT name, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
````

The inner query calculates the average salary. The outer query returns employees whose salary is above that value.

## 22. Common Table Expressions (CTEs)

### Step 1 – Understand the Concept

A CTE creates a temporary named result that can be referenced by the main query. It can make complex queries easier to understand.

### Step 2 – Demonstrate the Concept

Instead of placing an average-salary calculation directly inside a condition, I can first give the calculation a meaningful name such as `average_salary`.

### Step 3 – Implement the Concept

````python
WITH average_salary AS (
    SELECT AVG(salary) AS avg_salary
    FROM employees
)
SELECT e.name, e.salary
FROM employees AS e
CROSS JOIN average_salary AS a
WHERE e.salary > a.avg_salary;
````

The `WITH` clause creates the `average_salary` CTE. The main query then uses that result.

## 23. Views

### Step 1 – Understand the Concept

A view is a saved query that can be used like a virtual table. It generally stores the query definition rather than a separate copy of the underlying data.

### Step 2 – Demonstrate the Concept

If a company repeatedly needs a report containing employee names and departments, a view can save that query definition so users can query the view directly.

### Step 3 – Implement the Concept

````python
CREATE VIEW employee_department_view AS
SELECT
    e.name,
    d.department_name
FROM employees AS e
INNER JOIN departments AS d
    ON e.department_id = d.department_id;

SELECT *
FROM employee_department_view;
````

The `CREATE VIEW` statement saves the query. The view can then be queried using `SELECT`.

## 24. Indexes

### Step 1 – Understand the Concept

An index is an additional data structure that can help a database find rows faster for suitable queries. Indexes can improve read performance, but they also require storage and can add work to data modifications.

### Step 2 – Demonstrate the Concept

An index is similar to an index in a book. Instead of scanning every page to find a topic, the index helps locate the relevant pages faster.

### Step 3 – Implement the Concept

````python
CREATE INDEX idx_employees_department
ON employees(department_id);
````

The index is created on `department_id`, which can help suitable searches and joins involving that column.

# Part 5 – Window Functions

## 25. Window Functions – Introduction

### Step 1 – Understand the Concept

A window function performs a calculation across a related set of rows while keeping the individual rows in the result.

This is different from `GROUP BY`, which normally combines rows into one row per group.

### Step 2 – Demonstrate the Concept

Suppose I want to show every employee and also show their ranking within their department. I still need each employee's row, so a window function is useful.

### Step 3 – Implement the Concept

The examples below use:

```text
employees(employee_id, name, department, salary)
```

## 26. ROW_NUMBER()

### Step 1 – Understand the Concept

`ROW_NUMBER()` assigns a unique sequential number to rows within each window.

### Step 2 – Demonstrate the Concept

Within each department, employees can be numbered from highest salary to lowest salary.

### Step 3 – Implement the Concept

````python
SELECT
    name,
    department,
    salary,
    ROW_NUMBER() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS row_number
FROM employees;
````

`PARTITION BY department` creates a separate numbering group for each department. `ORDER BY salary DESC` determines the numbering order.

## 27. RANK()

### Step 1 – Understand the Concept

`RANK()` assigns the same rank to tied values. When ties occur, later ranks can have gaps.

### Step 2 – Demonstrate the Concept

If two employees have the same highest salary, both can receive rank 1. The next employee receives rank 3.

### Step 3 – Implement the Concept

````python
SELECT
    name,
    department,
    salary,
    RANK() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees;
````

Equal salary values receive the same rank, and the next rank can skip a number.

## 28. DENSE_RANK()

### Step 1 – Understand the Concept

`DENSE_RANK()` is similar to `RANK()`, but it does not leave gaps after tied values.

### Step 2 – Demonstrate the Concept

If two employees share rank 1, the next distinct salary receives rank 2 instead of rank 3.

### Step 3 – Implement the Concept

````python
SELECT
    name,
    department,
    salary,
    DENSE_RANK() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS dense_salary_rank
FROM employees;
````

Ties receive the same rank, but ranking continues without gaps.

## 29. LAG()

### Step 1 – Understand the Concept

`LAG()` allows a row to access a value from an earlier row in the selected window.

### Step 2 – Demonstrate the Concept

When looking at monthly sales, I may want to compare the current month's value with the previous month's value.

### Step 3 – Implement the Concept

Assume:

```text
monthly_sales(month_date, sales)
```

````python
SELECT
    month_date,
    sales,
    LAG(sales) OVER (
        ORDER BY month_date
    ) AS previous_month_sales
FROM monthly_sales;
````

`LAG(sales)` brings the previous row's sales value into the current row.

## 30. LEAD()

### Step 1 – Understand the Concept

`LEAD()` allows a row to access a value from a later row in the selected window.

### Step 2 – Demonstrate the Concept

For monthly sales, I can use `LEAD()` to see the next month's sales beside the current month's value.

### Step 3 – Implement the Concept

````python
SELECT
    month_date,
    sales,
    LEAD(sales) OVER (
        ORDER BY month_date
    ) AS next_month_sales
FROM monthly_sales;
````

`LEAD(sales)` brings the next row's sales value into the current row.

# 31. Combined Window Function Example

### Step 1 – Understand the Concept

Multiple window functions can be used together when different types of row-by-row analysis are required.

### Step 2 – Demonstrate the Concept

For employee salary analysis, I can show:
- Unique row numbering
- Ranking with gaps for ties
- Ranking without gaps for ties

### Step 3 – Implement the Concept

````python
SELECT
    name,
    department,
    salary,
    ROW_NUMBER() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS row_number,
    RANK() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS salary_rank,
    DENSE_RANK() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS dense_salary_rank
FROM employees;
````

The three window functions use the same department-based window but produce different numbering behavior.
