## 2. Reading Text Files

### Step 1 – Understand the Concept

Reading means bringing text stored in a file into the program so it can be displayed or processed. Python provides methods such as `read()`, `readline()`, and `readlines()` for different reading needs.

### Step 2 – Demonstrate the Concept

Reading a file is like opening a book: `read()` reads everything, `readline()` reads one line at a time, and `readlines()` collects the lines together.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
# Read the complete file

with open("sample_read.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)


with open("sample_read.txt", "r", encoding="utf-8") as file:
    first_line = file.readline()
    second_line = file.readline()

print("First line:", first_line.strip())
print("Second line:", second_line.strip())


with open("sample_read.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print(lines)
```

### Implementation Explanation

The examples use `open()` with read mode and demonstrate complete-file, single-line, and multiple-line reading.

## 3. Writing Text Files

### Step 1 – Understand the Concept

Writing puts new text into a file. Write mode replaces existing content, while append mode adds content to the end.

### Step 2 – Demonstrate the Concept

Write mode is like replacing the pages of a notebook, while append mode is like adding a new entry after the existing entries.

The example/analogy above shows the main idea before moving to the implementation.

### Implementation Explanation

The code creates a student file and then appends another line. The modes show the difference between replacing and extending file content.

### Step 2 – Demonstrate the Concept

It is like borrowing a library book inside a controlled checkout period: once the work is finished, the system handles returning it.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
with open("context_example.txt", "w", encoding="utf-8") as file:
    file.write("The with statement manages the file automatically.")

print("File operation completed.")


file = open("context_check.txt", "w", encoding="utf-8")
file.write("Example")
file.close()

print("File closed:", file.closed)

with open("context_check.txt", "r", encoding="utf-8") as file:
    print(file.read())

print("File closed after with block:", file.closed)
```

### Implementation Explanation

The examples use `with open(...)` and show that the file is closed after leaving the block.

## 5. CSV Files

### Step 1 – Understand the Concept

CSV is a simple tabular format where values are stored in rows and columns. Python's `csv` module provides tools for writing and reading this data.

### Step 2 – Demonstrate the Concept

A CSV file is like a spreadsheet saved as plain text: each row represents a record and each column represents a field.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import csv

students = [
    ["Name", "Age", "Marks"],
    ["Vignesh", 22, 85],
    ["Arun", 21, 78],
    ["Kumar", 23, 91]
]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("CSV file created.")


import csv

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)


import csv

records = [
    {"name": "Vignesh", "course": "Python", "score": 85},
    {"name": "Arun", "course": "Python", "score": 78}
]

with open("student_records.csv", "w", newline="", encoding="utf-8") as file:
    fieldnames = ["name", "course", "score"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(records)

with open("student_records.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

### Implementation Explanation

The code writes rows with `csv.writer`, reads them back, and also demonstrates dictionary-based CSV handling with `DictWriter` and `DictReader`.

## 6. JSON Files

### Step 1 – Understand the Concept

JSON stores structured data using objects, arrays, and simple values. It is useful when data needs to be saved or exchanged in a readable structured format.

### Step 2 – Demonstrate the Concept

JSON is like a labeled form: instead of only storing values, it stores each value with a key such as `name`, `age`, or `skills`.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import json

student = {
    "name": "Vignesh",
    "age": 22,
    "skills": ["Python", "C", "C++"],
    "course_completed": False
}

with open("student.json", "w", encoding="utf-8") as file:
    json.dump(student, file, indent=4)

print("JSON file created.")


import json

with open("student.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print(data)
print("Name:", data["name"])
print("Skills:", data["skills"])


import json

student = {"name": "Vignesh", "age": 22}

json_string = json.dumps(student)
print("JSON string:", json_string)

python_object = json.loads(json_string)
print("Python object:", python_object)
```

### Implementation Explanation

The code saves a Python dictionary as JSON, reads it back, and demonstrates converting between Python objects and JSON strings.

## 7. Pickle Files

### Step 1 – Understand the Concept

Pickle serializes Python objects into a binary representation so they can be saved and reconstructed later. Unlike JSON, it is Python-specific and can preserve more Python object types.

### Step 2 – Demonstrate the Concept

It is like packing a Python object into a storage box so the program can unpack it later in Python.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import pickle

student = {
    "name": "Vignesh",
    "age": 22,
    "skills": ["Python", "C", "C++"]
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

print("Pickle file created.")


import pickle

with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

### Implementation Explanation

The dictionary is serialized with `pickle.dump()` and reconstructed with `pickle.load()`. Pickle files should only be loaded from trusted sources.

## 10. shutil

### Step 1 – Understand the Concept

`shutil` provides higher-level file and directory operations such as copying and moving files.

### Step 2 – Demonstrate the Concept

It is like a moving assistant: instead of manually recreating a file, it can copy or move the existing item for you.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
import shutil
from pathlib import Path

source = Path("student.txt")
copy_path = Path("student_copy.txt")

if source.exists():
    shutil.copy2(source, copy_path)
    print("File copied:", copy_path)
else:
    print("Source file does not exist.")


import shutil
from pathlib import Path

source = Path("student_copy.txt")
destination = Path("student_moved.txt")

if source.exists():
    shutil.move(str(source), str(destination))
    print("File moved to:", destination)
else:
    print("Source file does not exist.")
```

### Implementation Explanation

The code copies a file with `copy2()` and moves it with `move()`.

## 11. File Validation

### Step 1 – Understand the Concept

File validation means checking whether a path points to an acceptable file before using it. Useful checks include existence, file type, extension, and size.

### Step 2 – Demonstrate the Concept

Before opening a package, I might check that it exists, is the expected type, and is not empty. File validation performs similar checks in software.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
from pathlib import Path

def validate_file(file_name, allowed_extensions=None):
    path = Path(file_name)

    if not path.exists():
        return False, "File does not exist."

    if not path.is_file():
        return False, "Path is not a file."

    if path.stat().st_size == 0:
        return False, "File is empty."

    if allowed_extensions and path.suffix.lower() not in allowed_extensions:
        return False, f"Invalid file type: {path.suffix}"

    return True, "File is valid."

valid, message = validate_file(
    "student.txt",
    allowed_extensions={".txt"}
)

print(valid, "-", message)


from pathlib import Path

csv_path = Path("students.csv")

if csv_path.exists() and csv_path.is_file():
    if csv_path.suffix.lower() == ".csv" and csv_path.stat().st_size > 0:
        print("CSV file is valid.")
    else:
        print("CSV file is invalid or empty.")
else:
    print("CSV file was not found.")
```

### Implementation Explanation

The validation function checks existence, whether the path is a file, whether it is empty, and whether its extension is allowed.

## 12. Directory Operations

### Step 1 – Understand the Concept

Directory operations include creating folders, creating nested folders, listing contents, creating files inside folders, and removing files or empty folders.

### Step 2 – Demonstrate the Concept

A directory is like a physical folder. I can create it, put files inside it, inspect its contents, and remove items when they are no longer needed.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
from pathlib import Path

directory = Path("file_handling_demo")

directory.mkdir(exist_ok=True)

print("Directory created:", directory)


from pathlib import Path

nested_directory = Path("file_handling_demo/data/raw")

nested_directory.mkdir(parents=True, exist_ok=True)

print("Nested directories created:", nested_directory)


from pathlib import Path

directory = Path("file_handling_demo")

for item in directory.iterdir():
    print(item)


from pathlib import Path

file_path = Path("file_handling_demo/data/raw/example.txt")
file_path.write_text("Example data", encoding="utf-8")

print("Created:", file_path)


from pathlib import Path

file_path = Path("file_handling_demo/data/raw/example.txt")

if file_path.exists():
    file_path.unlink()
    print("File removed.")
else:
    print("File does not exist.")


from pathlib import Path

empty_directory = Path("file_handling_demo/data/raw")

if empty_directory.exists():
    empty_directory.rmdir()
    print("Empty directory removed.")


import shutil
from pathlib import Path

demo_directory = Path("file_handling_demo")

if demo_directory.exists():
    shutil.rmtree(demo_directory)
    print("Demo directory removed.")
else:
    print("Demo directory does not exist.")
```

### Implementation Explanation

The examples demonstrate creating directories, nested directories, listing contents, creating a file, deleting a file, and removing an empty directory.

## 13. Directory Operations with `shutil`

### Step 1 – Understand the Concept

`shutil` can also remove an entire directory tree. This is powerful and should be used carefully because it can delete the directory and everything inside it.

### Step 2 – Demonstrate the Concept

It is like throwing away a physical folder together with all documents inside it. The operation should only be performed when the contents are no longer needed.

The example/analogy above shows the main idea before moving to the implementation.

### Step 3 – Implement the Concept



```python
from pathlib import Path
import json

records_directory = Path("student_records")
records_directory.mkdir(exist_ok=True)

record_path = records_directory / "student.json"

student = {
    "name": "Vignesh",
    "age": 22,
    "course": "Python",
    "skills": ["Python", "C", "C++"]
}

with record_path.open("w", encoding="utf-8") as file:
    json.dump(student, file, indent=4)

if record_path.exists() and record_path.is_file():
    print("Student record saved successfully.")
    print(record_path.read_text(encoding="utf-8"))
else:
    print("Student record validation failed.")
```

### Implementation Explanation

The code checks whether the demo directory exists and then removes it with `shutil.rmtree()`.
